# Ray Job Submission

- Initialise the Cluster Job Client
- Provide an entrypoint command directed to your job script
- Set up your runtime environment

```python
# Create authentication object for user permissions
# IF unused, SDK will automatically check for default kubeconfig, then in-cluster config
# KubeConfigFileAuthentication can also be used to specify kubeconfig path manually
from codeflare_sdk import get_cluster, TokenAuthentication

auth = TokenAuthentication(
    token = "sha256~XXX", # TODO: Replace with your own token
    server = "https://api.ocp.xxx.sandboxxxx.opentlc.com:6443", # TODO: Replace with your own server
    skip_tls=True
)
auth.login()
```

```python
# Get the cluster object
cluster = get_cluster(cluster_name="ray-cluster",namespace="xxx-ray-train", write_to_file=False) # TODO: Replace with your own cluster namespace
```

```python
# Initialize the Job Submission Client
"""
The SDK will automatically gather the dashboard address and authenticate using the Ray Job Submission Client
"""
client = cluster.job_client

# Submit an example mnist job using the Job Submission Client
submission_id = client.submit_job(
    entrypoint="python mnist_fashion.py",
    runtime_env={"working_dir": "./","pip": "requirements.txt"},
)
print(submission_id)
```

```python
# Get the status of the job
client.get_job_status(submission_id)
```
